This notebook contains scripts for analyzing house price data and generating reports and demonstrates using Squirrel in polyglot notebooks for data analysis and reporting.

```polyglot-notebook
#r "nuget: TableAPI, 1.0.4"
```

```polyglot-notebook
using Squirrel;

var data = DataAcquisition.LoadCsv("house.csv");
data.AddColumn(columnName:"price_per_sqm", formula:"[price]/[net_sqm]",decimalDigits:2);

string html = data.Pick("bedroom_count", "price_per_sqm")
.Aggregate("bedroom_count", AggregationMethod.Average)
.RoundOffTo(2)
//Parameters: string column, string ledgendText, string title, BarChartType chartType
.ToBarChartByGoogleDataVisualization("bedroom_count", "Price Per Sqm","Price", GoogleDataVisualizationcs.BarChartType.Column);

```

```polyglot-notebook
using System.IO;

File.WriteAllText("house-price-notebook.csv", data.ToCsv());

StreamWriter writer = new StreamWriter("house-price-notebook-chart.html");
writer.Write(html);
writer.Close();
```
